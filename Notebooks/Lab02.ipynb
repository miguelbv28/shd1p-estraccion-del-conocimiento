```python
import os
import pandas as pd

# Buscamos de forma segura si la carpeta es 'DataSet' o 'Data'
if os.path.exists('../DataSet'):
    ruta_data = '../DataSet'
elif os.path.exists('DataSet'):
    ruta_data = 'DataSet'
elif os.path.exists('../Data'):
    ruta_data = '../Data'
else:
    ruta_data = 'Data'

os.makedirs(ruta_data, exist_ok=True)
print(f" Variable 'ruta_data' definida correctamente como: {ruta_data}")
```

```python
# 1. Datos y creación de Netflix.csv
datos_netflix = {
    'show_id': ['s1', 's2', 's3', 's4', 's5'],
    'type': ['Movie', 'TV Show', 'Movie', 'Movie', 'TV Show'],
    'title': ['Dick Johnson Is Dead', 'Blood & Water', 'Ganglands', 'Jailbirds New Orleans', 'Kota Factory'],
    'director': ['Kirsten Johnson', None, 'Julien Leclercq', None, None],
    'country': ['United States', 'South Africa', 'France', None, 'India'],
    'release_year': [2020, 2021, 2021, 2021, 2021]
}
df_netflix = pd.DataFrame(datos_netflix)
df_netflix.to_csv(os.path.join(ruta_data, 'Netflix.csv'), index=False)

# 2. Datos y creación de Amazon.data
datos_amazon = {
    'id': ['am1', 'am2', 'am3', 'am4', 'am5'],
    'title': ['The Grand Tour', 'The Boys', 'Fleabag', 'Tom Clancy\'s Jack Ryan', 'Invincible'],
    'genre': ['Entertainment', 'Action', 'Comedy', 'Action', 'Animation'],
    'release_year': [2016, 2019, 2016, 2018, 2021]
}
df_amazon = pd.DataFrame(datos_amazon)
df_amazon.to_csv(os.path.join(ruta_data, 'Amazon.data'), index=False)

# 3. Datos y creación de Dataset.extension
datos_disney = {
    'id': ['dis1', 'dis2', 'dis3', 'dis4', 'dis5'],
    'title': ['The Mandalorian', 'WandaVision', 'Loki', 'Soul', 'Hamilton'],
    'type': ['TV Show', 'TV Show', 'TV Show', 'Movie', 'Movie'],
    'release_year': [2019, 2021, 2021, 2020, 2020]
}
df_disney = pd.DataFrame(datos_disney)
df_disney.to_csv(os.path.join(ruta_data, 'Dataset.extension'), index=False)

print(" ¡Excelente! Los tres archivos exigidos por la rúbrica se han guardado con éxito.")
```
